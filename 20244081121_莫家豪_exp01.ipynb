# 实验 2：函数、条件、循环与递归

### 2.1 `score_level(score)` 成绩等级

从高分段往低分段依次判断；不落在 0～100 范围时抛出 `ValueError`。

````python
def score_level(score):
    """90~100 优秀, 80~89 良好, 60~79 及格, 0~59 不及格; 超出范围抛 ValueError."""
    if not (0 <= score <= 100):
        raise ValueError(f"分数必须在 0~100 范围内, 收到 {score}")
    if score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"

for s in (95, 90, 85, 80, 70, 60, 30, 0):
    print(f"score_level({s:>3}) = {score_level(s)}")

for bad in (-5, 101):
    try:
        score_level(bad)
    except ValueError as e:
        print(f"score_level({bad:>3}) -> ValueError: {e}")
````

### 2.2 `for` 与 `while` 计算 1~100 的整数和

````python
total_for = 0
for i in range(1, 101):          # for 循环: 遍历 1..100
    total_for += i

total_while = 0
j = 1
while j <= 100:                  # while 循环: 手动控制计数器
    total_while += j
    j += 1

print("for   循环求和:", total_for)
print("while 循环求和:", total_while)
print("两种实现结果相同:", total_for == total_while)
````

### 2.3 递归版与迭代版阶乘

约定 `0! = 1`；输入必须是非负整数。递归版：`n! = n * (n-1)!`，基例 `0!, 1! = 1`。

````python
def factorial_recursive(n):
    """递归版: n! = n * (n-1)!, 基例 0! = 1."""
    if n < 0:
        raise ValueError("n 必须是非负整数")
    if n <= 1:
        return 1
    return n * factorial_recursive(n - 1)

def factorial_iterative(n):
    """迭代版: 从 2 乘到 n."""
    if n < 0:
        raise ValueError("n 必须是非负整数")
    result = 1
    for k in range(2, n + 1):
        result *= k
    return result

print(f"{'n':>3} {'递归版':>8} {'迭代版':>10}   是否一致")
for n in (0, 1, 5, 10):
    r1, r2 = factorial_recursive(n), factorial_iterative(n)
    print(f"{n:>3} {r1:>8} {r2:>10}   {r1 == r2}")
````

### 2.4 `summary(values)` 以元组返回统计值

返回 `(最小值, 最大值, 平均值)`，调用处通过**元组解包**一次性接收三个结果。

````python
def summary(values):
    """接收非空数值列表, 返回 (最小值, 最大值, 平均值) 元组."""
    if not values:
        raise ValueError("values 不能为空列表")
    return min(values), max(values), sum(values) / len(values)

lowest, highest, average = summary([85, 58, 92, 76, 60, 89])   # 元组解包
print(f"最小值 = {lowest}, 最大值 = {highest}, 平均值 = {average:.2f}")
````

### 2.5 `sum_all(*args)` 可变数量参数

`*args` 把所有位置参数打包成元组；不传参数时 `args` 为空元组，和为 0。

````python
def sum_all(*args):
    """返回所有参数之和."""
    return sum(args)

print("sum_all()               =", sum_all())                  # 无参数 -> 0
print("sum_all(7)              =", sum_all(7))                 # 单个参数
print("sum_all(1, 2, 3, 4, 5)  =", sum_all(1, 2, 3, 4, 5))      # 多个参数
````

### 3.5 引用赋值 vs `copy()`

- `alias = demo` 只复制**引用**，两个名字指向同一个列表对象；
- `copied = demo.copy()` 是**浅拷贝**，产生新的列表对象。

````python
demo = [10, 20, 30]                 # 单独的演示列表
alias = demo                        # 引用赋值: 不创建新列表
copied = demo.copy()                # 浅拷贝: 创建新列表

alias[0] = 999                      # 通过 alias 修改

print("demo   :", demo)              # 原列表跟着变了
print("alias  :", alias)            # 与 demo 是同一个对象
print("copied :", copied)           # 副本不受影响

print("alias is demo  :", alias is demo)
print("copied is demo :", copied is demo)
````

### 4.2 写入并读取 `students.txt`

`with open(..., encoding="utf-8")` 写入原始数据；再读回文件，跳过表头，用字符串方法 `split(",")` 解析每条记录。

````python
raw_lines = [
    "name,score",
    "Alice,85",
    "Bob,58",
    "Carol,92",
    "David,76",
    "Eva,60",
    "Frank,89",
    "Grace,abc",
    "Henry,105",
    "Incomplete",
]

students_path = os.path.join(data_dir, "students.txt")
with open(students_path, "w", encoding="utf-8") as f:      # 写入
    f.write("\n".join(raw_lines) + "\n")
print("已写入:", students_path)

with open(students_path, encoding="utf-8") as f:           # 读取
    lines = [line.rstrip("\n") for line in f]

parsed = []
for line in lines[1:]:                                     # 跳过表头
    name, _, score_str = line.partition(",")
    parsed.append((name, score_str))
print("解析结果:", parsed)
````

### 4.3 `parse_student(line)` 校验函数

校验项：字段数量必须为 2、姓名不能为空、成绩能转成整数且在 0～100 范围内。
任何一条不满足都抛出 `ValueError`。

````python
def parse_student(line):
    """解析一行 '姓名,成绩' 文本; 校验失败抛出 ValueError."""
    fields = line.split(",")
    if len(fields) != 2:
        raise ValueError(f"字段数量错误: 期望 2 个, 实际 {len(fields)} 个")
    name, score_str = fields[0].strip(), fields[1].strip()
    if not name:
        raise ValueError("姓名为空")
    try:
        score = int(score_str)
    except ValueError:
        raise ValueError(f"成绩不是整数: {score_str!r}") from None
    if not (0 <= score <= 100):
        raise ValueError(f"成绩超出 0~100 范围: {score}")
    return {"name": name, "score": score}

print(parse_student("Alice,85"))     # 正常记录
````

### 4.4 容错式逐行处理

捕获每行的 `ValueError`，记录**行号与错误原因**后继续处理后续记录；
有效记录以字典形式保存进列表。

````python
valid_records = []
errors = []

with open(students_path, encoding="utf-8") as f:
    header = f.readline()                       # 跳过表头
    for lineno, line in enumerate(f, start=2):  # 数据从第 2 行开始
        line = line.strip()
        if not line:
            continue
        try:
            valid_records.append(parse_student(line))
        except ValueError as e:
            errors.append((lineno, str(e)))     # 记录行号与原因, 继续处理

print("--- 无效记录 ---")
for lineno, msg in errors:
    print(f"第 {lineno} 行: {msg}")

print(f"\n--- 有效记录 ({len(valid_records)} 条) ---")
for record in valid_records:
    print(record)
````

### 4.5 写入 `report.txt`：`w` 与 `a` 模式

- `w` 模式：文件**截断为零**再写入，已有内容被完全覆盖；
- `a` 模式：在文件**末尾追加**，已有内容保留。

````python
report_path = os.path.join(data_dir, "report.txt")
n_valid = len(valid_records)
n_invalid = len(errors)
avg_score = sum(r["score"] for r in valid_records) / n_valid
pass_rate = sum(1 for r in valid_records if r["score"] >= 60) / n_valid

# ---- w 模式: 覆盖写 ----
with open(report_path, "w", encoding="utf-8") as f:
    f.write("=== 学生成绩处理报告 ===\n")
    f.write(f"有效记录数: {n_valid}\n")
    f.write(f"无效记录数: {n_invalid}\n")
    f.write(f"平均分    : {avg_score:.2f}\n")
    f.write(f"及格率    : {pass_rate:.2%}\n")

print("--- 第一次以 w 模式写入后 ---")
print(open(report_path, encoding="utf-8").read())

# ---- 再次以 w 模式写入: 旧内容被覆盖 ----
with open(report_path, "w", encoding="utf-8") as f:
    f.write("这一行覆盖了旧内容\n")

print("--- 再次以 w 模式写入(旧内容被覆盖) ---")
print(open(report_path, encoding="utf-8").read())

# ---- a 模式: 追加写, 保留已有内容 ----
with open(report_path, "a", encoding="utf-8") as f:
    f.write("=== 追加的统计 ===\n")
    f.write(f"有效 {n_valid} 条 | 无效 {n_invalid} 条 | 平均分 {avg_score:.2f} | 及格率 {pass_rate:.2%}\n")

print("--- 以 a 模式追加后(原内容保留) ---")
print(open(report_path, encoding="utf-8").read())
````

# 实验 5：类、继承与对象复制

### 5.1 定义 `Student` 类

`__init__` 初始化 `name` 与 `score`；`is_passed()` 判断是否及格；`__str__` 定义打印形式。

````python
class Student:
    """表示一名学生: 姓名与成绩."""

    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        return f"Student(name={self.name!r}, score={self.score}, 及格={self.is_passed()})"

demo_student = Student("Alice", 85)
print(demo_student)
print(demo_student.is_passed())
````

### 5.2 为实验 4 的每条有效记录创建 `Student` 对象

````python
students = [Student(record["name"], record["score"]) for record in valid_records]

for stu in students:
    print(f"{stu}   is_passed -> {stu.is_passed()}")
````

### 5.3 `GraduateStudent` 继承

- 通过 `super().__init__()` 复用父类初始化逻辑；
- 新增 `research_topic` 属性；
- 重写 `__str__` 补充研究方向；
- 子类实例同时也是父类的实例（`isinstance` 验证）。

````python
class GraduateStudent(Student):
    """研究生: 在 Student 基础上增加研究方向."""

    def __init__(self, name, score, research_topic):
        super().__init__(name, score)           # 调用父类 __init__
        self.research_topic = research_topic

    def __str__(self):                           # 重写 __str__
        return (f"GraduateStudent(name={self.name!r}, score={self.score}, "
                f"及格={self.is_passed()}, 研究方向={self.research_topic!r})")

gs = GraduateStudent("Alice", 85, "数据挖掘")
print(gs)
print("is_passed 继承自父类:", gs.is_passed())
print("isinstance(gs, Student):", isinstance(gs, Student))
````

### 5.4 直接赋值、`copy.copy()` 与 `copy.deepcopy()`

- **直接赋值**：不创建新对象，两个名字指向同一个对象；
- **`copy.copy()` 浅拷贝**：外层容器是新的，但内部嵌套的列表仍然**共享**；
- **`copy.deepcopy()` 深拷贝**：递归复制所有层级，完全独立。

每次比较都从新建的原始对象开始。

````python
import copy

# ---- 1) 直接赋值: 同一对象 ----
o1 = {"name": "Alice", "scores": [85, 90]}
r1 = o1
r1["scores"].append(95)
print("直接赋值    : 修改副本后原对象 =", o1, "<- 被同步修改")

# ---- 2) copy.copy: 内层列表共享 ----
o2 = {"name": "Alice", "scores": [85, 90]}
r2 = copy.copy(o2)
r2["scores"].append(95)
print("copy.copy   : 修改副本后原对象 =", o2, "<- 内层列表仍共享, 被同步修改")

# ---- 3) copy.deepcopy: 完全独立 ----
o3 = {"name": "Alice", "scores": [85, 90]}
r3 = copy.deepcopy(o3)
r3["scores"].append(95)
print("copy.deepcopy: 修改副本后原对象 =", o3, "<- 不受影响")
````

## 小结

| 概念 | 关键结论 |
|------|----------|
| 运算符 | `/` 恒为 float，`//` 向下取整，`%` 求余，`**` 幂 |
| 字符串 | 不可变；`strip/split/join/切片` 是数据处理基本功 |
| 函数 | `*args` 打包位置参数；元组可一次返回多个值再解包 |
| 容器 | `=` 是引用，`.copy()` 浅拷贝；元组不可变；`zip` 取短 |
| 文件与异常 | `with open` 自动关闭；`w` 覆盖、`a` 追加；捕获异常保证批处理不中断 |
| 类与复制 | 继承复用父类逻辑；嵌套结构要用 `deepcopy` 才真正独立 |
