## 热身：定义简单函数

1. `say_hi()`：无参数，打印 `"Hello, world!"`；
2. `goat_pad(s)`：打印字符串 `s`，前后都加上 `"goat"`；
3. `print_n(s, n)`：把字符串 `s` 打印 `n` 次，每次单独一行。

```python
def say_hi():
    """打印 Hello, world!"""
    print("Hello, world!")

def goat_pad(s):
    """打印前后都加上 goat 的字符串."""
    print("goat" + s + "goat")

def print_n(s, n):
    """把字符串 s 打印 n 次, 每次单独一行."""
    for _ in range(n):
        print(s)

say_hi()
goat_pad("milk")
print_n("go", 3)
```

## 欧几里得算法

最大公约数（GCD）：`gcd(m, n)` 反复执行 `m, n = n, m % n`，直到 `n` 为 0，此时 `m` 即最大公约数。

```python
def gcd(m, n):
    """欧几里得算法求最大公约数."""
    while n != 0:
        m, n = n, m % n
    return m

print("gcd(2018, 2019) =", gcd(2018, 2019))
print("gcd(1200, 300)  =", gcd(1200, 300))
print("gcd(5040, 60)   =", gcd(5040, 60))
```

### 2.3 负数参数的行为

直接把负数传入上面的实现，观察结果。

```python
print("gcd(-12, 8)   =", gcd(-12, 8))
print("gcd(12, -8)   =", gcd(12, -8))
print("gcd(-12, -8)  =", gcd(-12, -8))
```

**讨论**：Python 里 `%` 的结果符号跟随**除数**，所以上述实现遇到负数时可能返回负的"gcd"（例如 `gcd(12, -8)` 返回 `-4`）。
这在数学上没有意义——最大公约数按定义应为**正整数**。若要支持负数输入，应在开头取绝对值：
`m, n = abs(m), abs(n)`，再执行欧几里得迭代。

## 近似欧拉数 e

自然对数的底数 e 有两种经典近似：

- **极限式**：`e ≈ (1 + 1/x)^x`，当 `x → ∞`；
- **无穷级数**：`e = Σ(k=0..∞) 1/k! = 1 + 1 + 1/2 + 1/6 + 1/24 + ...`，约定 `0! = 1`。

注意 `euler_infinite_sum` 的参数 `n` 是 **1-indexed** 的项数：`n=1` 只取级数第 1 项（k=0 项）返回 1；`n=0` 返回 0。

```python
import math

def euler_limit(n):
    """用 (1 + 1/x)^x 取 x = n 近似 e."""
    return (1 + 1 / n) ** n

def euler_infinite_sum(n):
    """级数前 n 项之和(1-indexed): euler_infinite_sum(0) = 0, (1) = 1."""
    return float(sum(1 / math.factorial(k) for k in range(n)))

def euler_approx(epsilon):
    """不断累加级数项, 直到近似值与真实 e 的误差小于 epsilon."""
    total = 0.0
    term = 1.0          # 当前项 1/k!, 从 k = 0 开始
    k = 0
    while abs(total - math.exp(1)) >= epsilon:
        total += term
        k += 1
        term /= k       # 1/k! = 1/(k-1)! / k
    return total

def print_euler_sum_table(n):
    """打印 euler_infinite_sum(k), k 从 1 到 n, 每行一个."""
    for k in range(1, n + 1):
        print(euler_infinite_sum(k))
```

### 测试

```python
# 1) euler_limit: n 越大越接近 math.e
print("math.e          =", math.e)
for n in (1, 10, 100, 100000):
    print(f"euler_limit({n:<6}) =", euler_limit(n))

print()

# 2) euler_infinite_sum: 0 -> 0, 1 -> 1, 4 -> 1+1+1/2+1/6 ≈ 2.667
print("euler_infinite_sum(0) =", euler_infinite_sum(0))
print("euler_infinite_sum(1) =", euler_infinite_sum(1))
print("euler_infinite_sum(4) =", euler_infinite_sum(4), "≈ 2.667")

print()

# 3) euler_approx: 误差小于 epsilon
approx = euler_approx(1e-6)
print("euler_approx(1e-6) =", approx)
print("与 math.e 的误差   =", abs(math.e - approx), "< 1e-6")

print()

# 4) print_euler_sum_table
print("print_euler_sum_table(10):")
print_euler_sum_table(10)
print("math.e =", math.e, "<- 逐行逼近")
```

## 测试整数属性

正整数 `n` 是 2 的幂，当且仅当 `n = 2^p`（`p ≥ 0`）。
判断方法：**反复用 2 整除 `n`**，若最终恰好得到 1，则是 2 的幂（只用了除法与 `%`，不涉及 `math.sqrt`）。

```python
def is_power_of_2(n):
    """n 是 2 的幂返回 True(只需除法和取模)."""
    if n < 1:
        return False
    while n % 2 == 0:
        n //= 2
    return n == 1

def is_power(b, n):
    """is_power(b, n): n 是 b 的幂返回 True.
    n = 1 时恒为 True(任意底数的 0 次幂); b = 1 时只有 n = 1 是幂."""
    if n == 1:
        return True
    if b < 2:
        return False
    while n % b == 0:
        n //= b
    return n == 1

print("is_power_of_2 测试:")
for n in (1, 2, 3, 4, 7, 8, 16, 31, 64, 100):
    print(f"  is_power_of_2({n:>3}) = {is_power_of_2(n)}")

print()
print("is_power 测试:")
for b, n in ((2, 8), (3, 81), (4, 8), (5, 125), (2, 18), (5, 1), (1, 1), (1, 7)):
    print(f"  is_power({b}, {n:>3}) = {is_power(b, n)}")
```
